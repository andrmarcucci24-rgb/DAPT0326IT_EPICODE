```python
## Esercizio 2 - Operatori e precedenza
prezzo = 45.90
quantita = 3
sconto_percentuale = 10
nome = 'Mario'
cognome = 'Rossi'

# 1. calcolare il totale dell'ordine (prezzo*quantita) e poi applicare lo sconto, usando le parentesi per far calcolare prima il totale e poi la percentuale
tot_ordine = (prezzo * quantita) - ((prezzo * quantita) * 0.10)
print(tot_ordine)

# 2. costruire la stringa saluto concatenando nome e cognome con l'operatore
print("Ciao" + " " + nome + " " + cognome)

# 3. usare un operatore di assegnazione composto per aggiornare la variabilie quantita 
quantita += 1
print(quantita)

# un operatore è un simbolo che indica di eseguire una determinata operazione 
# / restituisce il risultato completo di una divisione 
# // restituisce solo la parte intera del risultato della divisione
# % restituisce il resto 

# es. 
divisione = 12/5
intero = 12//5
resto = 12%5
print(divisione)
print(intero)
print(resto)
```

```python
# Esercizio 3 - liste e tuple
prezzi = [5, 10, 15, 20, 25] # lista
coordinate = (36.6120, 7.4321) # tupla

# 1. sulla lista prezzi aggiungere un nuovo prezzo con append() e ordinarla con sort()
prezzi.append(22)
prezzi.sort()
print(prezzi)

# 2. estrarre il primo e l'ultimo prezzo della lista usando indexing
print(prezzi[0])
print(prezzi[5])
print(prezzi[-6])
print(prezzi[-1])

# 3. provare a modificare il primo valore sulla tupla e riportare l'erroe
coordinate[0] = 40.7895

# TypeError: 'tuple' object does not support item assignment -> questo erroe perché le tuple non sono modificabili 
# lista: è un insieme di dati ordinati che ammette modifiche e duplicati
# tupla: è un insieme di dati ordinati che non può essere modicato e ammette duplicati 
```

```python
# Esercizio 4 - set e dizionario
lista_voti = [7, 8, 7, 9, 8, 10, 7] # lista
rubrica = {  # dizionario
        "Anna" : "1234",
        "Marco" : "5678"
}

# 1. ottenere l'insime di voti unici a partire dalla lista e calcolare il numero di duplicati 
voti_unici = set(lista_voti)
print(voti_unici)

numero_duplicati = len(lista_voti) - len(voti_unici)
print(numero_duplicati)

# 2. aggiungere alla rubrica un nuovo contatto e aggiornare il numero di un contatto già presente, usando l'assegnazione tramite chiave
rubrica["Andrea"] = "9999" # aggiungo tramite chiave che ancora non esiste a rubrica 
print(rubrica)

# 3. usare get() per cercare un contatto che non esiste in rubrica, confrontando il risultato con quello ottenuto con l'altro 
print(rubrica.get("Federica"))  # None # non restituisce errore anche se non trova la chiave
print(rubrica["Federica"]) # resituisce errore

# Dizionario: è strutturato in un rapporto chiave - valore, è modificabile, l'accesso è tramite chiave e le chiavi per questo motivo non ammettono duplicati
```

```python
# Esercizio 5 - Slicing e metodi combinati
# dati: 
vendite = [120, 340, 275, 410, 90, 560, 205]
testo = " Analisi Vendite Trimestre "

# 1- su vendite eseguire separatente vendite[::2] e vendite[::-1]
print(vendite[::2]) # stampa un valore ogni 2 partendo dal primo valore della lista che è 0 
print(vendite[::-1]) # stampa tutti i valori partendo dall'ultimo

# 2- eseguire separatamente vendite[:3] e vendite[:4]
print(vendite[:3]) # stampa da 0 a 2 il 3 è escluso
print(vendite[:4]) # stessa cosa ma da 0 a 3

# 3- su testo applicare in sequenze strip(), lower(), split() sul risultato precedente e stampare tutto 
print(testo.strip())  # rimuove spazi vuoti o caratteri specifici all'inizio e alal fine di una stringa
testo_lower = testo.lower() # scrive tutto in minuscolo
testo_lower
print(testo_lower.split()) # divide le parole di una stringa unica 

# lo slicing permette di estrarre una sottosequenza di un oggetto seqeunziale come una lista o una tupla 
```
